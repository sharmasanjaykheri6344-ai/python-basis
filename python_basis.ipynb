Q.2 Explain the difference between List and Tuple in Python?

...>> In Python, both List and Tuple are used to store collections of items, but they differ in important ways:

🔹 1. Mutability
List → Mutable (can be changed after creation)
Example: You can add, remove, or modify elements.
Tuple → Immutable (cannot be changed after creation)
Once created, its values stay fixed.
🔹 2. Syntax
List → Uses square brackets [ ]
Example: my_list = [1, 2, 3]
Tuple → Uses parentheses ( )
Example: my_tuple = (1, 2, 3)
🔹 3. Performance
List → Slightly slower (due to flexibility and mutability)
Tuple → Faster (because it is fixed and optimized)
🔹 4. Use Case
List → Used when data may change (e.g., dynamic data, user inputs)
Tuple → Used when data should remain constant (e.g., fixed records, coordinates)
🔹 5. Methods Available
List → Many methods (append(), remove(), sort(), etc.)
Tuple → Limited methods (count(), index())
✅ Example:
# List (mutable)
my_list = [1, 2, 3]
my_list.append(4)   # Allowed

# Tuple (immutable)
my_tuple = (1, 2, 3)
# my_tuple.append(4)  # Error
🔸 Summary:
Use List when you need flexibility
Use Tuple when you need safety and fixed data>

Q.3 🔹 What is a Function in Python?

A function in Python is a reusable block of code that performs a specific task. It is defined using the def keyword and can be called whenever needed in a program.

Example:

def greet():
    print("Hello, World!")

greet()  # calling the function
🔹 Why are Functions Useful?
Code Reusability
You can write a function once and use it multiple times, avoiding repetition.
Better Organization
Functions help break large programs into smaller, manageable parts, making code easier to read and understand.
Easy Maintenance
If you need to update logic, you only change it in one place (inside the function), instead of everywhere in the code.
Improves Debugging
Errors can be easily identified and fixed since functions isolate specific tasks.
Supports Modularity
Functions allow you to build programs in a structured and modular way.
🔸 Summary:

A function is a reusable block of code, and it is useful because it makes programs simpler, cleaner, and more efficient.

Q.5  Write a Python program to check whether a number is even or odd using conditional
statements?
..>> Here’s a simple Python program to check whether a number is even or odd using conditional statements:

# Take input from user
num = int(input("Enter a number: "))

Q.8 Write a Python program to remove duplicate values from a list using a set?

...>> Here’s a simple Python program to remove duplicate values from a list using a set:

# Original list with duplicates
numbers = [1, 2, 3, 2, 4, 1, 5]

# Remove duplicates using set
unique_numbers = list(set(numbers))

# Print result
print("List after removing duplicates:", unique_numbers)
🔹 How it works:
A set automatically removes duplicate values
set(numbers) converts the list into a set (unique elements only)
list() converts it back to a list
🔸 Output:
List after removing duplicates: [1, 2, 3, 4, 5].

Q.10 Write a Python program to count how many times a number appears in a list.

Example List

[2,3,4,2,5,2]

..>> Here’s a simple Python program to count how many times a number appears in a list:

# Given list
numbers = [2, 3, 4, 2, 5, 2]

count() method counts how many times the element appears in the list
User inputs the number to search
🔸 Example Output:
